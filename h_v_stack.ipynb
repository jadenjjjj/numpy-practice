题 5：判断能不能 stack
A.shape = (4, 5)
B.shape = (4, 6)

不能，stack 的 shape必须要完全一样

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

np.vstack([a, b]).shape （2，3）
np.hstack([a, b]).shape  （6，）
np.stack([a, b], axis=0).shape （2，3）
np.stack([a, b], axis=1).shape   （3， 2）

```python
a = np.ones((1, 3))
b = np.ones((1, 2))
```

np.vstack([a, b]) 不行，因为是垂直堆积，增加的是行，列要一样。
np.hstack([a, b]) 可以，水平叠加，增加的是列，行要一样
np.stack([a, b]) 不行，shape必须一模一样

```python
a = np.ones((2, 3))
b = np.ones((2, 3))
```

np.vstack([a, b]).shape  4，3
np.hstack([a, b]).shape 2，6
np.stack([a, b], axis=0).shape 2 2 3
np.stack([a, b], axis=1).shape 2 2 3
np.stack([a, b], axis=2).shape 2 3 2

hstack 全是一维数组时，结果还是一维。vstack可以自动将一维变成二维
