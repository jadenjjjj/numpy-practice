```python
import numpy as np

roux = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

gumbo = np.array([
    [10, 20],
    [30, 40]
])
```

```python
roux.shape
```

```python
gumbo.shape
```

题 1：上下拼 roux 和 gumbo

```python
result1 = np.concatenate([roux,gumbo], axis=0)
result1
```

题 2：左右拼 roux 和 gumbo
不能拼，左右拼看行 3，2 不同

题 3：roux 和 roux 上下拼

```python
result2 = np.concatenate([roux,roux], axis = 0)
result2
```

```python
result2.shape
```

题 4：roux 和 roux 左右拼

```python
result3 = np.concatenate([roux, roux], axis = 1)
result3
```

```python
result3.shape
```

```python
a = np.ones((2, 3))
b = np.ones((2, 5))
```

```python
a.shape
```

```python
b.shape
```

np.concatenate([a, b], axis=0)

不能 列不同， 3/5

np.concatenate([a, b], axis=1)

可以，左右，增加列数， 最后shape是（2，8）



```python
c = np.ones((4, 3))
d = np.ones((2, 3))
```

```python
c.shape
```

```python
d.shape
```

np.concatenate([c, d], axis=0)

可以，增加的是行数，最后是 （6，3）

np.concatenate([c, d], axis=1)

不可以，增加的是列数，行数要一样， 4 2 不同

```python
x = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

y = np.array([
    [10, 20, 30]
])
```

题 9：创建并拼接 要求：把 y 拼到 x 的下面。

```python
result9 = np.concatenate([x,y], axis = 0)
```

```python
result9
```

```python
result9.shape
```

题 10：把一列拼到右边, 要求：把 z 拼到 x 的右边。

```python
x = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

z = np.array([
    [100],
    [200]
])
```

```python
result10 = np.concatenate([x,z], axis=1)
```

```python
result10
```

```python
result10.shape
```

题 11：三个数组上下拼

```python
a = np.array([
    [1, 2]
])

b = np.array([
    [3, 4],
    [5, 6]
])

c = np.array([
    [7, 8]
])
```

```python
result11 = np.concatenate([a,b,c], axis=0)
```

```python
result11.shape
```

最后补一个容易犯错的点：1D 数组和 2D 数组不能随便拼, 维度看外层括号，或者ndim

```python
result11.ndim
```
