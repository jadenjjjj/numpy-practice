当然。下面是这节 **4.1 `where()`** 的整理版笔记，按“你以后复习能看懂”的方式写。

**NumPy 4.1 `where()` 笔记**

这一节讲的是：

```text
用 np.where() 代替 for 循环里的 if/else。
```

也就是：当你想对 NumPy 数组里的每个元素做判断时，不要一个一个循环，尽量让 NumPy 一次性处理。

核心公式：

```python
np.where(condition, value_if_true, value_if_false)
```

中文意思：

```text
如果 condition 为 True，就用 value_if_true；
如果 condition 为 False，就用 value_if_false。
```

例子：

```python
import numpy as np

foo = np.array([10, 20, 30, 40])
bar = np.array([1, 2, 3, 4])

baz = np.where(bar % 2 == 0, foo * 2, foo / 2)
baz
```

结果：

```python
array([ 5., 40., 15., 80.])
```

解释：

```text
bar = 1，是奇数，所以 foo = 10 / 2 = 5
bar = 2，是偶数，所以 foo = 20 * 2 = 40
bar = 3，是奇数，所以 foo = 30 / 2 = 15
bar = 4，是偶数，所以 foo = 40 * 2 = 80
```

普通慢写法是：

```python
baz = np.zeros(foo.size)

for i in range(foo.size):
    if bar[i] % 2 == 0:
        baz[i] = foo[i] * 2
    else:
        baz[i] = foo[i] / 2
```

这段代码没错，但是慢。因为 Python 是一轮一轮处理，每次循环都要和底层 C 代码来回沟通。

NumPy 快写法是：

```python
baz = np.where(bar % 2 == 0, foo * 2, foo / 2)
```

这叫 **vectorization，向量化**。

简单理解：

```text
for 循环：一个一个算
np.where：整批一起算
```

所以 NumPy 会快很多。

视频里对比速度：

```text
for 循环：大约 2.17 秒
np.where：大约 57 毫秒
```

`57 毫秒 = 0.057 秒`，所以 `np.where()` 快非常多。

视频还用了：

```python
np.array_equal(baz1, baz2)
```

作用是检查两个数组是否完全一样。

例如：

```python
np.array_equal(
    np.array([1, 2, 3]),
    np.array([1, 2, 3])
)
```

结果：

```python
True
```

在视频里，它用来证明：

```text
for 循环算出来的结果
```

和：

```text
np.where() 算出来的结果
```

是一样的，只是 `np.where()` 更快。

测速用的是 Jupyter 里的：

```python
%%timeit
```

它可以测一段代码的运行时间。

比如：

```python
%%timeit
np.where(bar % 2 == 0, foo * 2, foo / 2)
```

**本节最重要的记忆点**

看到这种逻辑：

```python
for i in range(...):
    if 条件:
        做 A
    else:
        做 B
```

如果处理的是 NumPy 数组，就优先想：

```python
np.where(条件, A, B)
```

你要背下来的句子：

```python
np.where(condition, yes, no)
```

意思是：

```text
条件成立选 yes，不成立选 no。
```

我们刚才额外练到的补充知识：

```python
(condition1) & (condition2)
```

表示“并且”。

```python
(condition1) | (condition2)
```

表示“或者”。

注意多个条件一定要加括号：

```python
(amount >= 100) | is_member
```

不要写成：

```python
amount >= 100 | is_member
```

这个容易算错。

下一步我们就严格回到这节课内容来练：  
**把 for + if/else 改写成 np.where()**。这才是 4.1 的核心能力。
