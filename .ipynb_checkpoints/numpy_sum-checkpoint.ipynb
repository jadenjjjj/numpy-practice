```python
import numpy as np

squee = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
```

题 1：全部求和
要求：把所有数字加起来。

```python
result1 = np.sum(squee)
result1
```

题 2：按列求和

```python
result2 = np.sum(squee, axis = 0)
result2
```

题 3：按行求和

```python
result3 = np.sum(squee, axis = 1)
result3
```

题 4：按列求和，但保留二维

```python
result4 = np.sum(squee, axis=0, keepdims=True)
result4.shape
```

题 5：按行求和，但保留二维

```python
result5 = np.sum(squee, axis=1, keepdims=True)
result5
```

```python
result5.shape
```

```python
squee = np.array([
    [1., 2., 3.],
    [4., 5., 6.]
])
```

```python
squee[0,0] = np.nan
squee
```

```python
result6 = np.sum(squee)
result6
```

```python
result7 = np.nansum(squee)
result7
```

```python
result8 = np.sum(np.nan_to_num(squee))
result8
```

题 1：全部求和

```python
import numpy as np

arr = np.array([
    [2., 4., 6.],
    [1., 3., 5.]
])
```

```python
result9 = np.sum(arr)
result9
```

题 2：按列求和

```python
result10 = np.sum(arr, axis=0)
result10
```

题 3：按行求和

```python
result11 = np.sum(arr, axis=1)
result11
```

```python
result11.shape
```

题 4：按列求和，保留二维

```python
result12 = np.sum(arr, axis=0, keepdims=True)
result12
```

```python
result12.shape
```

题 5：按行求和，保留二维

```python
result13 = np.sum(arr, axis=1, keepdims=True)
result13
```

```python
result13.shape
```

```python
arr_nan = np.array([
    [np.nan, 4., 6.],
    [1., 3., 5.]
])
```

```python
result14 = np.nansum(arr_nan)
result14
```

题 8：用 np.nansum() 按列求和

```python
result15 = np.nansum(arr_nan, axis=0)
result15
```

题 9：用 np.nansum() 按行求和

```python
result16 = np.nansum(arr_nan, axis=1)
result16
```

题 10：全部销售额

```python
sales = np.array([
    [100., 200., np.nan],
    [50., np.nan, 150.],
    [80., 120., 100.]
])
```

```python
result17 = np.nansum(sales)
result17
```

题 11：每个商品总销售额

```python
result18 = np.nansum(sales, axis=0)
result18
```

题 12：每天总销售额

```python
result19 = np.nansum(sales, axis=1)
result19
```
