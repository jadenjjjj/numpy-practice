```python
np.isnan(foo)
```

题 1：检查每个元素是不是 NaN

```python
foo = np.array([
    [1., 2., 3.],
    [np.nan, 5., 6.],
    [np.nan, np.nan, np.nan],
    [7., 8., 9.]
])
```

```python
is_nan = np.isnan(foo)
is_nan
```

题 2：找出“至少有一个 NaN 的行”

```python
result2 = np.any(np.isnan(foo), axis=1)
result2
```

```python
foo[result2]
```

题 3：找出“全部都是 NaN 的行”

```python
result3 = np.all(np.isnan(foo), axis=1)
result3
```

```python
foo[result3]
```

题 4：找出“没有任何 NaN 的行”

```python
result4 = ~np.any(np.isnan(foo), axis=1)
result4
```

```python
foo[result4]
```

题 5：找出“不是全部 NaN 的行”

```python
result5 = ~np.all(np.isnan(foo), axis=1)
result5
```

```python
foo[result5]
```

题 6：找出“至少有一个 NaN 的列”

```python
result6 = np.any(np.isnan(foo), axis=0)
result6
```

```python
foo[:,result6]
```

```python
baz = np.array([
    [1., np.nan, 3., np.nan],
    [5., np.nan, 7., np.nan],
    [9., np.nan, 11., np.nan]
])
result8 = np.all(np.isnan(baz), axis=0)
result8
```

题 9：筛出没有任何 NaN 的行

```python
data = np.array([
    [1., 2., 3.],
    [np.nan, 5., 6.],
    [7., 8., 9.],
    [np.nan, np.nan, np.nan]
])
result9 = ~np.any(np.isnan(data), axis=1)
result9
```
